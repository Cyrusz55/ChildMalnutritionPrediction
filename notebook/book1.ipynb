```python
df = df.copy()

df["has_malnutrition"] = (
    (df["haz_height_for_age_zscore_x100"] < -200)
    | (df["whz_weight_for_height_zscore_x100"] < -200)
    | (df["waz_weight_for_age_zscore_x100"] < -200)
).astype(int)
```
